## Задание 2 - МАТРИЧНЫЕ И ВЕКТОРНЫЕ ОПЕРАЦИИ
Вычислите определитель, ранг, норму, вектор собственных значений матрицы G и матрицу, обратную ей:
$$
G = \begin{pmatrix}
2 & 3+\alpha & 4+\mu & 6 \\
\beta+1 & \gamma & \nu & 12 \\
5 & -2 & 16 & \mu+1 \\
2 & 7 & 11 & -6
\end{pmatrix}
$$


```python
import numpy as np

# Значения параметров (также будут использоваться дальше в №№3.1, 3.2, 4)
alpha = np.random.randint(0, 25)
beta  = np.random.randint(0, 25)
gamma = np.random.randint(0, 25)
mu    = np.random.randint(0, 25)
nu    = np.random.randint(0, 25)

print(f"Сгенерированные alpha, beta, gamma, mu, nu:")
print(alpha, beta, gamma, mu, nu)
print("\n")

# Матрица G 
G = np.array([
    [2, 3 + alpha, 4 + mu, 6],
    [beta + 1, gamma, nu, 12],
    [5, -2, 16, mu + 1],
    [2, 7, 11, -6]
], dtype=float)

# Определитель 
det = np.linalg.det(G)

# Ранг 
rank = np.linalg.matrix_rank(G)

# Нормы
norm_fro = np.linalg.norm(G, 'fro')      # норма Фробениуса
norm_2   = np.linalg.norm(G, 2)          # спектральная норма
norm_1   = np.linalg.norm(G, 1)          # столбцовая норма
norm_inf = np.linalg.norm(G, np.inf)     # строчная норма

# Собственные значения и векторы 
eigvals, eigvecs = np.linalg.eig(G)

# Обратная матрица 
if abs(det) > 1e-12:
    G_inv = np.linalg.inv(G)
    check = G @ G_inv
    inv_report = f"Обратная матрица:\n{G_inv}\n\nПроверка G @ G⁻¹ (должна быть единичная):\n{check}"
else:
    inv_report = "Матрица вырождена — обратной не существует."

# Итог
print("Матрица G:\n", G)
print(f"\nОпределитель det(G) = {det:.6f}")
print(f"Ранг rank(G)         = {rank}")
print(f"\nНорма Фробениуса     = {norm_fro:.6f}")
print(f"Спектральная норма   = {norm_2:.6f}")
print(f"Норма ||G||₁         = {norm_1:.6f}")
print(f"Норма ||G||∞         = {norm_inf:.6f}")
print("\nСобственные значения:\n", eigvals)
print("\nСобственные векторы (по столбцам):\n", eigvecs)
print("\n" + inv_report)
```

### Задание 3.1
Решите систему линейных уравнений согласно варианту.
$$
\left\{
\begin{array}{rcl}
3x_1 + x_2 + x_3 + \gamma x_4 & = & \alpha \\
x_1 - \mu x_2 + \nu x_3 + 4x_4 & = & \beta \\
-5x_1 - x_3 - 7x_4 & = & -5 \\
x_1 - 6x_2 + \alpha x_3 + 6x_4 & = & 3
\end{array}
\right.
$$

```python
left = np.array([
    [3, 1, 1, gamma], 
    [1, -mu, nu, 4],
    [-5, 0, -1, -7],
    [1, -6, alpha, 6]
])

right = np.array([alpha, beta, -5, 3]) # матрица из правой части (решения)

result = np.linalg.solve(left, right)  # решение 

for i in range(1, 5):
    print(f"x{i} = {result[i-1]:.3f}")
```

### Задание 3.2
Найдите решение системы линейных алгебраических уравнений методом наименьших квадратов:
$$
\left\{
\begin{array}{rcl}
3x_1 + x_2 + x_3 + \gamma x_4 & = & \alpha \\
x_1 - \mu x_2 + \nu x_3 + 4x_4 & = & \beta \text{ .} \\
-5x_1 - x_3 - 7x_4 & = & -5
\end{array}
\right.
$$

```python
left = np.array([
    [3, 1, 1, gamma],
    [1, -mu, nu, 4],
    [-5, 0, -1, -7]
])

right = np.array([alpha, beta, -5])

# решение, сумма квадратов невязок, ранг, сингулярные числа
result, residuals, rank, s = np.linalg.lstsq(left, right, rcond=None)

for i in range(1, 5):
    print(f"x{i} = {result[i-1]:.3f}")
print(f"\nРанг матрицы (left): {rank}")
print(f"Сумма квадратов невязок: {residuals}")
```

## Задание 4 - СТАТИСТИЧЕСКАЯ ОБРАБОТКА ДАННЫХ
1. Сгенерируйте массив из n = 500 + 10μ + ν чисел, распределенных по нормальному закону с математическим ожиданием, равным ν и дисперсией, равной μ + 2.

2. Разбейте интервал значений на k подынтервалов. Количество подынтервалов выберите по правилу Стерджеса: k = 1 + 3.322*lg(n). Значение k округлите в большую сторону. Подсчитайте абсолютные и относительные частоты попадания в каждый интервал.

3. Результаты представьте в виде датафрейма, столбцы которого должны содержать следующую информацию: номер интервала, левый конец интервала, его середина, правый конец, абсолютные и относительные частоты попадания в интервал.

4. Подсчитайте накопленную сумму частот (функция cumsum()), среднее значение элементов массива, медиану, выборочную дисперсию (смещенную и несмещенную), среднеквадратическое отклонение и 30 % квантиль. Поясните полученные результаты.

```python
# Задание параметров и генерация массива
n = int(np.ceil(500 + 10 * mu + nu))
print(f"Объем выборки n = {n}")

# Генерация: loc = мат. ожидание (nu), scale = СКО (sqrt(mu+2))
arr = np.random.normal(loc=nu, scale=np.sqrt(mu + 2), size=n)

# Разбиение на интервалы (k = [6; 12] обычно)
k = int(np.ceil(1 + 3.322 * np.log10(n)))
print(f"Количество подынтервалов k = {k}\n")

xmin, xmax = arr.min(), arr.max()          # мин. и макс. значения
edges = np.linspace(xmin, xmax, k + 1)     # k+1 границ для k интервалов
freq, _ = np.histogram(arr, bins=edges)    # абсолютные частоты
rel_freq = freq / n                        # относительные частоты
midpoints = (edges[:-1] + edges[1:]) / 2   # середины интервалов
cum_freq = np.cumsum(freq)                 # накопленные частоты

# Формирование датафрейма (можно было с помощью pandas)
table_list = []
for i in range(k):
    table_list.append({
        'Номер интервала': i + 1,
        'Левый конец': edges[i],
        'Середина': midpoints[i],
        'Правый конец': edges[i + 1],
        'Абсолютная частота': freq[i],
        'Относительная частота': rel_freq[i],
        'Накопленная частота': cum_freq[i]
    })

print("ТАБЛИЦА РАСПРЕДЕЛЕНИЯ:")
print(f"{'№':<3} {'Левый':>10} {'Середина':>10} {'Правый':>10} "
      f"{'Абс.част':>10} {'Отн.част':>10} {'Накопл.':>8}")
print("-" * 67)

for row in table_list:
    print(f"{row['Номер интервала']:<3} "
          f"{row['Левый конец']:>10.4f} "
          f"{row['Середина']:>10.4f} "
          f"{row['Правый конец']:>10.4f} "
          f"{row['Абсолютная частота']:>10d} "
          f"{row['Относительная частота']:>10.4f} "
          f"{row['Накопленная частота']:>8d}")

# Расчёт стат. характеристик и вывод
mean_val = np.mean(arr)
median_val = np.median(arr)
var_biased = np.var(arr)           # Смещенная дисперсия (ddof=0)
var_unbiased = np.var(arr, ddof=1) # Несмещенная дисперсия (ddof=1)
std_dev = np.std(arr)              # СКО (смещенное)
quantile_30 = np.quantile(arr, 0.30)

print("\nСТАТИСТИЧЕСКИЕ ХАРАКТЕРИСТИКИ:")
print(f"Среднее значение (mean):         {mean_val:.4f}")
print(f"Медиана (median):                {median_val:.4f}")
print(f"Смещенная дисперсия:             {var_biased:.4f}")
print(f"Несмещенная дисперсия            {var_unbiased:.4f}")
print(f"Среднеквадратическое отклонение: {std_dev:.4f}")
print(f"30% квантиль:                    {quantile_30:.4f}")
```
