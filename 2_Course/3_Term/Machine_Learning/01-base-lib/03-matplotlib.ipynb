# Сгенерируйте случайную матрицу из 1000 точек размерности 2

 

```python
points = np.random.randint(0, 100, size=(1000, 2))
```

# Нарисуйте сгенерированные в п.3 точки на плоскости

```python
plt.scatter(points[:, 0], points[:, 1])

plt.grid(True, linestyle='--')
plt.show()
```

# постройте гистограммы первого и второго столбца матрицы из п.3, расположите их на двух картинках размещённых в один ряд (subplots)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(points[:, 0])
axes[1].hist(points[:, 1])

plt.show()
```

# нарисуйте график функции $\frac{sin(x^2)}{x}$

```python
x = np.linspace(-20, 20, 300)

def f(x):
    return np.where(x==0, 0, np.sin(x**2) / x)

plt.plot(x, f(x))
plt.grid(True, linestyle='--')
plt.show()
```
