# Как добавить границу (заполненную нулями) вокруг существующего массива?

```python
# с помощью функции pad

arr = np.random.randint(1, 100, size=(5, 5))
print("Исходный массив:")
print(arr)

arr = np.pad(arr, pad_width=1, mode='constant', constant_values=0)
print("\nС границей из нулей:")
print(arr)
```

# Сгенерируйте два случайных массива A и B, проверьте их на равенство

 

```python
a = np.random.randint(-25, 25, size=10)
b = np.random.randint(-25, 25, size=10)

print("A:", a)
print("B:", b)

print("\nПоэлементно")
print(a == b)

print("\nРавенство массивов")
print(np.array_equal(a, b))
```

# Сгенерируйте случайную матрицу размером 10х2, представляющую декартовы координаты, преобразуйте их в полярные координаты

 

```python
dec = np.random.uniform(-10, 10, size=(10, 2))

x = dec[:, 0]
y = dec[:, 1]

print("Декартовы координаты (x, y):")
print(dec)

r = np.sqrt(x**2 + y**2)
theta = np.arctan2(y, x)

polar = np.column_stack((r, theta))

print("\nПолярные координаты (r, theta в радианах):")
print(polar)
```

# Для двух случайных массивов X и Y постройте матрицу C 
$C_{ij} = \frac{1}{x_i - y_j}$

```python
x = np.random.rand(4)
y = np.random.rand(4)

print("X:", x)
print("Y:", y)
print()

c = 1 / (x[:, None] - y[None, :])

print(c)
```

# Сгенерируйте две случайные матрицы A и B, перемножьте их

```python
a = np.random.randint(1, 10, size=(5, 5))
b = np.random.randint(1, 10, size=(5, 5))

print("A:\n", a)
print("B:\n", b)
print()

result = a @ b

print(result)
```
