# Сгенерируйте вектор случайных значений numpy размером 1000, создайте из него DataFrame, имя колонки 'x'

```python
import numpy as np

x = np.random.rand(1000)

df = pd.DataFrame({'x': x})

df
```

# Посчитайте минимальное и мексимальное значение столбца 'x', а также медиану и квартили

```python
maxVal = df['x'].max()
minVal = df['x'].min()
median = df['x'].median()

q1 = df['x'].quantile(0.25) # 25% квантиль
q2 = df['x'].quantile(0.5)  # 50%
q3 = df['x'].quantile(0.75) # 75%

print(f"Минимум:  {minVal:.4f}")
print(f"Максимум: {maxVal:.4f}")
print(f"Медиана:  {median:.4f}")
print(f"Q1 (25%): {q1:.4f}")
print(f"Q2 (50%): {q2:.4f}")
print(f"Q3 (75%): {q3:.4f}")
```

# Нарисуйте график плотности распределения 'х'

```python
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(df, x='x')

plt.xlabel('Значение x', fontsize=12)
plt.ylabel('Плотность вероятности', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

# Нарисуйте блочную диаграмму с ограничителями выбросов ("ящик с усами") для 'х'

```python
sns.boxplot(y=df['x'])

# 4. Оформление
plt.title('Блочная диаграмма ("ящик с усами") для столбца "x"', fontsize=14)
plt.ylabel('Значение', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

# Отфильтруйте все значения меньше 0.8 для 'х'

```python
df_less = df[df['x'] < 0.8]

print("Оставлены значения < 0.8")
df_less
```
