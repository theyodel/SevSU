# `**kwargs` (сокращение от KeyWord ARGuments)
Расскажем об одном способе передачи аргументов функции,
использующем словари. Напомним два способа передачи аргументов:

```python
def myfunc(x=0, y=1): # функция с параметрами x=0 и y=0 (по умолчанию)
    print("x =", x)
    print("y =", y)
```

```python
myfunc(12, 19)
```

```python
myfunc(y=2, x=5)
```

Как показывает второй пример, аргументы можно передавать,
указывая их имена. Допустим, мы хотим написать функцию, которая
принимает неопределенное количество именованных аргументов (мы
даже не знаем заранее, каких). Это можно сделать следующим
образом:


```python
def new_func(**kwargs): # функция, которая принимает на вход kwarg'и (бишь именные аргументы)
    print(kwargs)
```

```python
new_func(x=1, y=5, z=8, s = "Some string") # на выходе работы функции получаем словарь, который состоит по следующему прототипу: key - имя аргумента, value - значение аргумента
```

Две звёздочки в определении функции `new_func()` говорят
следующее: «все именованные параметры, переданные этой функции,
следует поместить в словарь `kwargs`». Как видим, именно так это и
работает: например, параметр `x=1` превратился в запись `'x': 1` в
словаре `kwargs`. Однако, это происходит только с «бесхозными»
параметрами: если бы у функции был отдельный параметр `x`, то он
бы не попал в `kwargs`. Например:


```python
def other_func(x, **kwargs): # функция принимает на вход первым аргументом аргумент x, а потом - именные аргументы
    print(kwargs)
other_func(x=10,y=5)
```

# Сортировка
Сортировка — то есть расположение элементов списка в каком-то
определённом порядке — распространённая программистская задача.
Для сортировки списков в Python есть два главных инструмента.
Первый — метод `sort()`, выполняющий сортировку *in place*, то есть
внутри самого списка. Например:

```python
my_list = [6, 9, 2, 7, 12, 8]
```

```python
my_list.sort() # сортируем исходный список "внутри него самого" (бишь "на месте") с помощью метода .sort()
```

```python
my_list
```

Метод `sort()` *меняет* исходный список (и поэтому, кстати, умеет
работать только со списками — у кортежей такого метода нет). Если
вместо этого вы хотите создать новый список, то следует использовать
функцию `sorted()`.

```python
my_list = [6, 9, 2, 7, 12, 8]
```

```python
sorted_list = sorted(my_list) # сортируем список в новый с помощью функции sorted()
sorted_list
```

Так мы создали новый список. Старый при этом остался без
изменений.


```python
my_list # исходный список остался без изменений, т.к. функция sorted() не меняет исходный список. Она возвращает уже отсортированный список
```

Функцию `sorted()` можно применять не только к спискам, но и к
неизменяемым последовательностям — например, к кортежам. На
выходе всегда получается список.

```python
my_tuple = (7, 1, 2, 6)
print(sorted(my_tuple)) # функция sorted() применяется не только к спискам, но и к неизменяемым типам, таким как кортежи
```

## Сортировка строк
Сортировать можно списки, состоящие не только из чисел, но и из
более сложных объектов — лишь бы умели сравнивать их между
собой. Например, строки можно сравнивать между собой — они
упорядочены в *лексикографическом порядке*, то есть «по алфавиту» и
так, как они шли бы в словаре (имеется в виду обычный бумажный
словарь, а не тип данных Python).


```python
"abcd" < "b" # сравнение строк происходит лексикографически, т.е. по алфавиту
```

```python
"abcd" < "addd"
```

```python
"a" < "aa"
```

Вот так выглядит сортировка списка из строк:

```python
str_list = ["Bob", "Alice", "Bill", "Weigu"]
str_list.sort() # сортироовка списка строк происходит аналогично спискам чисел
str_list
```

## Сортировка и циклы
Можно использовать функцию `sorted()` совместно с оператором
`for` чтобы обрабатывать элементы списка в каком-то определённом
порядке. Например, у нас есть словарь и мы хотим вывести его
элементы в порядке возрастания значения ключа. Тогда об этом надо
явно попросить Python, как показывает следующий пример:

```python
gradebook = {'Bob': 3, 'Alice': 5, 'Weigu': 4, 'Bill': 2}
for k in gradebook:
    print(k, gradebook[k])
```

Как видите, элементы не упорядочены. Вот так их можно упорядочить
при выводе:


```python
for k in sorted(gradebook): # в данном случае функция sorted() сортирует словарь по ключам
    print(k, gradebook[k])
```

## Более сложные примеры сортировки
Можно сортировать список в обратном порядке (по убыванию). Для
этого используется параметр `reverse`.

```python
sorted([4, 8, 1, 7], reverse=True) # в данном случае функция sorted() принимает аргумент reverse со значением True, что влият на вывод отсортированного списка, а именно "разворачивает" отсортированный список
```

Можно сортировать не только числа и строки, но и более сложные
объекты. Нарпимер, рассмотрим такую табличку (реализованную как
список кортежей), в которой записаны имена студентов и их оценки по
нескольким работам.


```python
names = [("Bob", 8, 4, 9),
         ("Alice", 7, 8, 9),
         ("Weigu", 7, 5, 3),
         ("Dan", 6, 4, 3)]
```

```python
names.sort() # здесь сортировка списка из кортежей происходит по первому элементу каждого кортежа (первый элемент по умолчанию)
```

```python
names
```

Судя по результату, логично предположить, что сортировка была
выполнена по первому элементу — имени студента. Действительно,
кортежи сравниваются примерно так же, как строки,
лексикографически. Сначала сравниваются первые элементы:

```python
('a', 8) < ('b', 7)
```

Если первый элемент совпадает, то сравниваются вторые элементы и
т.д.

```python
('a', 8) < ('a', 7)
```

А что делать, если бы нам захотелось отсортировать кортежи в списке
names не по первому элементу, а по второму или ещё какому-нибудь?
Для этого необходимо использовать параметр `key`, задающий ключ
сортировки. Прежде, чем мы это сделаем, нужно сказать несколько
слов о том, как можно одной функции передать другую функцию в
качестве параметра.

## Отступление: функции как аргументы
Рассмотрим такую функцию:

```python
def superfunc(f): # функция superfunc() принимает в себя другую функцию и передаёт в неё аргументом число 2
    return f(2)
```

В качестве аргумента она принимает какую-то функцию `f`, вызывает
эту функцию передаёт ей в качестве аргумента число 2 и возвращает
тот результат, который вернула `f`.

Например:

```python
from math import sqrt
superfunc(sqrt) # вот здесь мы видим, что функция superfunc() приняла в себя функцию sqrt()
```

Мы импортировали функцию `sqrt()` из модуля `math`, после чего
передали функции `superfunc()` функцию `sqrt` в качестве
параметра. Обратите внимание: при передаче после функции `sqrt`
нет скобок: это потому, что мы её не *вызываем*, а *передаём* другой
функции. Функция `superfunc` взяла нашу функцию `sqrt` и
вызвала её, передав ей число 2 в качестве параметра. То есть
вычислила корень из двух.
Можете представить себе, что `sqrt` — это рецепт, записанный на
бумажке. Мы передаём его в виде такой бумажки функции
`superfunc` и она его как-то использует. Передадим другую бумажку
— она использует её. Например:

```python
def plusodin(x):
    return x + 1
```

```python
superfunc(plusodin)
```

Если мы попробуем передать функции `superfunc` что-то другое —
например, строчку или число — ничего не получится (она ожидает
именно функцию).

```python
superfunc("sqrt") # функция выдала ошибку, т.к. ждёт в качестве аргумента функцию, а не какую-то строку
```

## Ключи сортировки
Вернёмся к задаче о сортировке таблицы, представленной в виде
списка кортежей. Чтобы отсортировать такой список по второму
элементу, необходимо сначала создать функцию, которая будет
возвращать второй элемент переданного ей кортежа (или списка).

```python
def get_second_element(x): # функция get_second_element() возвращает второй элемент какого-либо списка/кортежа
    return x[1]
```

Посмотрим, как она работает:

```python
get_second_element([7, 8, 4, 2])
```

Теперь мы передаем эту функцию в качестве параметра `key` методу
`sort()` (функция `sorted()` тоже сработает):

```python
names.sort(key=get_second_element) # сортируем список names, который состоит из кортежей по вторым элементам кортежей. Для этого используем параметр key, в который передаём ключ, по которому мы и будем сортировать список
```

```python
names
```

Видно, что теперь строки оказались упорядочены по второму столбцу
(первой оценке): у `Dan` она самая низкая ( `6` ), у `Bob` самая высокая
( 8 ), а у `Alice` и `Weigu` одинаковые ( `7` ).

Возникает естественный вопрос: а как упорядочены в этом случае
строчки, соответствующие `Alice` и `Weigu`? Ответ: в том порядке, в
котором они шли в исходном списке. Это бывает удобно, если вы
хотите упорядочить сначала по одному параметру, а потом по другому:
просто выполните сортировку последовательно, сначала по второму
параметру, а потом по первому.

Чтобы не определять каждый раз функцию типа
`get_second_element` можно использовать готовую: для этого надо
импортировать специальную функцию `itemgetter`:


```python
from operator import itemgetter
```

```python
sorted(names, key=itemgetter(2)) # функция itemgetter() позволяяет нам не прописывать постоянно функццию на подобии get_second_element()
# В данном случае мы упорядочили по третьему столбцу
```

```python
sorted(names, key=itemgetter(3))
# А здесь упорядочили по четвертому столбцу
```

Допустим, мы хотим упорядочить по третьему столбцу, а если третий
столбец даёт одну и ту же оценку, то по алфавиту. Это можно сделать
так: *сначала* упорядочим по алфавиту, а *потом* — по третьему
столбцу.


```python
print(names)
names.sort(key=itemgetter(0)) # упорядочили по первому столбцу
print(names)
names.sort(key=itemgetter(2)) # упорядочили по третьему столбцу
print(names)
```

Больше подробностей о сортировке можно найти в [официальном
tutorial](https://docs.python.org/3/howto/sorting.html), а мы перейдём к
следующей теме.


# Форматирование строк
Зачастую требуется вставить значение каких-то переменных в какуюто строчку. Пример, который нам уже встречался.

```python
name = "Alice"
grade = 5
print("Student", name, "has grade", grade) # форматируем строки методом поочерёдной подстановки параметров, бишь чередуем строку и переменную, которую хотим вставить
```

С помощью `print()` можно вывести такую строку на печать, но
если бы мы хотели передать её какой-то другой функции, то надо было
бы придумывать что-то другое. И это другое уже придумано!

Есть два распространённых способа подставлять значение
переменных в строку (это часто называется *интерполяцией*, хотя не
имеет никакого отношения к одноименной математической операции).
Первый способ более классический.


```python
new_str = "Student %s has grade %i" % (name, grade) # В данном примере мы форматируем строку с помощью оператора %. В местах, где мы хотим вставить какое-то значение какой-то переменной мы пишем % и указываем тип вставляемых данных, ширину, расположение и т.д. (подобно printf() в Си)
print(new_str)
```

Здесь используется оператор `%`, который для строк делает
следующую операцию: берёт строку слева от него, находит там все
«поля для подстановки» (*placeholders*) — в данном случае это `%s` и
`%i`, после чего берёт переменные, перечисленные справа от него (это
может быть одна переменная или кортеж из нескольких переменных,
как в данном случае) и подставляет их последовательно — первую
переменную на место первого placeholder, вторую на место второго и
т.д.

Буквы в обозначениях placeholders означают тип переменной: в
данном случае `%s` — это строка, а `%i` — целое число. Вот ещё
несколько примеров:

```python
print("The number is %i" % 2.3)
print("The number is %f" % 2) # f значит float
print("The number is %.2f" % 2.1393) # два знака после точки
print("The number is %04i" % 3) # дополнить до четырёх знаков
```

При использовании оператор `%` нужно быть осторожным: он имеет
приоритет по сравнению с арифметическими операциями, поэтому вы
можете получить неожиданный результат, если не поставите скобки:

```python
print("a = %i" % 3*3) # здесь происходит "недопонимание". Питон вместо подставновки 9 печатает нашу строку 3 раза, ведь оператор * для строк просто повторяет её определённое кол-во раз
```

Здесь произошло следующее: сначала выполнился код `"a = %i" % 3`, а потом результат умножился на 3 (что для строчек эквивалентно
трёхкратному повторению). Если вы хотели подставить результат
выполнения `3*3`, то нужно было сделать вот так:


```python
print("a = %i" % (3*3)) # здесь мы уладили конфликт путём распределения очерёдности выполнения операций, а именно поставили скобки для той операции, которую хотим выполнить первой
```

Второй способ форматирования («новый») заключается в
использовании метода `format()`. Он действует примерно так:

```python
"hello, {0}, this is {1}, again {0}, {var}".format(7, 9, var="James") # второй способ форматирования строк - метод .format(), в котором мы указываем подставляемые значения.
# Работает он так: на место, куда хотим поставить какое-либо значение ставим {} и внутри фигурных скобок ставим индекс элемента из .format(), который хотим туда подставить
```

Здесь не приходится явно указывать типы данных (подставляется
строковое представление переменной). Одно и то же значение может
использоваться несколько раз (к ним можно обращаться по номерам и
именам). Впрочем, можно и не указывать явно номера — тогда
переменые будут подставляться по очереди:


```python
"Fist var: {}, the second one: {}".format(8, 1) # если в фигурных скобках не указывать индекс элемента, то туда будет подставляется всё по-очереди
```

Форматирование может быть довольно сложным и никто не в силах
запомнить все тонкости. Неплохая документация на эту тему (как по
оператору `%`, так и по методу `format()`) собрана [здесь](https://pyformat.info/).

Третий, самый современный способ — f-строки (f-strings, появился
начиная с Python версии 3.6). Перед строкой ставится буква `f`, а
внутри фигурных скобок пишутся имена переменных или любые выражения:

```python
print(f"Student {name} has grade {grade}") # Ну и наконец третий, и, наверное, самый распространённый способ форматирования строк - f-строки
# работают f-строки так: перед форматируемой строкой ставим f (например f'hello') и на местах, где необходимо что-либо подставить пишем {}, а в фигурных скобках - либо название переменной, либо значение, либо какую-лиюо операцию
```

```python
a = 5
b = 7
print(f"{a} + {b} = {a + b}")
```

Внутри `{}` можно вызывать методы, обращаться к элементам
списков, ключам словарей и даже выполнять небольшие вычисления:

```python
student = {"name": "Alice", "grade": 5}
print(f"{student['name']}: {student['grade']}")
```

Можно также задавать формат вывода чисел, например, количество
знаков после запятой:

```python
pi = 3.1415926
print(f"Pi = {pi:.4f}") # также внутри {} в f-строках можно применять функции форматирования, а именно: указывать точность после ',', длину строки и т.д.
```

f-строки работают быстрее, чем `.format()`, и делают код гораздо
более читаемым. Сегодня это рекомендуемый способ форматирования
строк в Python.

Форматирование может быть довольно сложным и никто не в силах
запомнить все тонкости. Неплохая документация на эту тему (как по
оператору `%`, так и по методу `format()` и f-строкам) собрана [здесь](https://pyformat.info/).

Кстати:

```python
print("%f" % (0.1+0.2))
```

Вроде бы, ничео неожиданно, но давайте увеличим точность…

```python
print("%.18f" % (0.1+0.2)) # по идее, должно было получиться 0.3(0), но т.к. в памяти ЭВМ мы записываем все числа в двоичной системе, поэтому и возникают некоторые ошибки при сравнении двух дробей
```

Когда мы попросили вывести результат с точностью 18 знаков после
запятой, откуда-то взялись непонятные значащие цифры в конце. Это
связано с тем фактом, что компьютеры используют двоичную систему
счисления, а в ней числа типа `0.1` записываются в виде *бесконечной*
периодической дроби и не могут быть представлены в виде конечной
дроби. При арифметических операциях возникают ошибки
округления, которые и приводят к таким эффектам.

Иногда эти эффекты становятся опасными. Вы думаете, что `0.1 + 0.2` это `0.3`? У вашего компьютера другое мнение на этот счёт:

```python
0.1 + 0.2 == 0.3 # вот пример ошибки
```

Однако, не стоит отчаиваться: вы можете использовать обыкновенные
дроби или специальный модуль `decimal` для работы с десятичными
дробями.

```python
from fractions import Fraction
```

```python
Fraction(1,10) + Fraction(2, 10) # фиксим ошибкуметодом ввода в нашу программу функции Fraction(), которая позволяет нам записывать числа в дробном формате. Первый аргумент - числитель, второй - знаменатель
```

```python
Fraction(1, 3) + Fraction(1, 2) # на выходе аналогично получаем дроби, записанные в виде функции Fraction()
```

```python
from decimal import Decimal
```

```python
Decimal("0.1") + Decimal("0.2") # второй способ исправления прикола с дробями - импорт функции Decimal(), которая принимает в себя дроби в виде строк (например "0.2")
```

```python
Decimal("0.1") + Decimal("0.2") == Decimal("0.3") # на выходе аналогично получаем дроби в виде строк
```

Подробнее о десятичных и бинарных дробях можно прочитать в
[официальной документации](https://docs.python.org/3/tutorial/floatingpoint.html#tut-fp-issues).

# Практические задачи
1. Напишите функцию, которая принимает произвольное количество
именованных аргументов и выводит каждый из них в формате
"ключ: значение" с помощью цикла `for`.


```python
def firstFunc(**kwargs):
    for k, v in kwargs.items():
        print(f'{k}: {v}')

firstFunc(name="Arsenii", age=18, course=1, university="SevSU")
```

2. Дан список `["Иван", "Мария", "Анна", "Дмитрий", "Елена"]`, отсортируйте его в обратном алфавитном порядке.

```python
names = ["Иван", "Мария", "Анна", "Дмитрий", "Елена"]
names.sort(reverse=True)
print(names)
```

```python
cities = [("Москва", 12), ("Сочи", 25), ("Казань", 8), ("Новосибирск", -15)]
cities.sort(key=itemgetter(1))
print(cities)
```

убыванию и сохраните результат в новый список.

6. Напишите функцию, которая принимает произвольное количество
именованных аргументов (например, `function(name="Анна", city="Москва", hobby="чтение", age=30)`) и выводит
только те аргументы, значение которых длиннее 5 символов (если
значение — строка), или больше 10 (если значение — число).

```python
def sixthFunc(**kwargs):
    for v in kwargs.values():
        if type(v)==str:
            if len(v)>5:
                print(v)
        elif type(v)==int or type(v)==float:
            if v>10:
                print(v)

sixthFunc(name="Julia", age=18, id=5, quote="Hello Wolrd!")
print("----------------------------------------------------")
sixthFunc(name="Анна", city="Москва", hobby="чтение", age=30)
```

7. Дан список строк `["арбуз", "дыня", "ананас", "яблоко", "киви"]`, отсортируйте его по последней букве каждого слова.

```python
fruits = ["арбуз", "дыня", "ананас", "яблоко", "киви"]
print(sorted(fruits, key=itemgetter(-1)))
```

8. Создайте строку `"Уважаемый Степан! Ваш заказ № 123 готов"`, подставив имя `"Степан"` и номер заказа `123`,
используя старый формат через оператор `%`.

```python
myStr = "Уважаемый %s! Ваш заказ № %d готов" % ("Степан", 123)
print(myStr)
```

9. Дан список `[10, 20, 30, 40, 50]`, отсортируйте его по
остатку от деления на 6.

```python
def rem6(x):
    return x%6

nums = [10, 20, 30, 40, 50]
nums.sort(key=rem6)

print(nums)
```

10. Дан словарь `{"Анна": 4, "Борис": 5, "Виктор": 3, "Галина": 5}`, выведите ключи словаря (имена) в
отсортированном по алфавиту порядке, а для каждого имени
выведите оценку в формате "Анна получила 4".

```python
gradebook = {"Анна": 4, "Борис": 5, "Виктор": 3, "Галина": 5}

for name in sorted(gradebook.keys()):
    print(f'{name} получил {gradebook[name]}')
```
