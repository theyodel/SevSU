# Словари
Рассмотрим такую задачу: у нас есть информация об оценках студентов
по некоторому предмету и мы хотим иметь возможность с этой
информацией работать — например, по имени студента определить,
какую оценку он получил. Мы могли бы пытаться решить эту задачу,
создав два списка — один с именами студентов, а другой с оценками:

```python
students = ["Вася", "Коля", "Петя", "Аня"]
grades = [5, 4, 2, 3]
```

Было бы хорошо, если бы у нас была возможность иметь тип данных, в
котором элементы нумеруются не натуральными числами, а
произвольными объектами. Такой тип данных существует: в Python он
называется *словарём* (*dictionary*).

Вот так можно создать словарь в Python:

```python
gradebook = {"Вася": 5, "Коля": 4, "Петя": 2, "Аня": 3} # создаём словарь, где в качестве ключа лежит имя студента (srting), а значение - его оценка (int)
```

Это похоже на создание списка, но есть ряд отличий. Во-первых, мы
использовали фигурные скобки вместо квадратных, чтобы показать, что
создаём именно словарь. Во-вторых, словарь состоит из *записей*, каждая
запись состоит из двух частей: *ключа* (*key*) и *значения* (*value*). Ключ и
значение разделяются двоеточием. Например, у нас есть запись `"Аня": 3` с ключом `"Аня"` и значением `3`. Всего наш словарь `gradebook`
сейчас содержит четыре записи, ключами которых являются имена
студентов, а значениями — их оценки.


```python
gradebook
```

Заметим, что при печати Python переупорядочил записи в словаре. На
самом деле, порядок вывода записей в словаре является произвольным:
внутри словаря записи не имеют никакого порядка. Поэтому нельзя
обратиться, например, к «первой записи», но зато можно обратиться к
записи с данным ключом:

```python
gradebook['Аня'] # получаем значение по ключу 'Аня'
```

```python
gradebook['Вася'] # получаем значение по ключу 'Вася'
```

Можно изменить значение записи, точно так же, как изменить элемент
списка.

```python
gradebook['Аня'] = 5 # изменяем значение по ключу 'Аня' на 5
```

```python
gradebook
```

Можно добавить новую запись.

```python
gradebook['Иннокентий'] = 4 # добавляем новую запись с ключом 'Иннокентий' и значением 4
```

```python
gradebook
```

При попытке обратиться к записи, которой нет, мы получим сообщение
об ошибке:


```python
gradebook['Alice'] # обращаемся к записи с ключом 'Alice', но получаем ошибку т.к. такого ключа не найдено
```

Часто нам хочется иметь возможность запросить запись, а в случае, если
её нет, получить какое-нибудь «значение по умолчанию», а не ошибку.
Для этого нужно использовать метод `get()` вместо квадратных скобок.


```python
gradebook.get('Alice') # Теперь при желании заполучить запись, которая не сущестует, наш код не будет прерываться, а просто выведет какое-либо значение по-умолчанию
```

Здесь вернулось `None`:

```python
print(gradebook.get('Alice'))
```

```python
gradebook.get('Вася')
```

Можно было бы передать `get()` второй аргумент, и тогда в случае,
если такого ключа в словаре нет, то будет возвращен он.


```python
gradebook.get('Alice', 'No such student')
```

```python
gradebook.get('Вася', "No such student")
```

Можно получить список всех ключей словаря:


```python
gradebook.keys() # получаем список из всех включей нашего словаря
```

На самом деле это не совсем список, но эта штука ведёт себя почти как
список и из неё можно сделать список. Аналогично со списком всех
значений словаря.


```python
gradebook.values() # получаем список из всех значений нашего словаря
```

Ключами словарей могут быть не только строчки. Допустим, мы хотим
создать словарь, в котором ключами будут числа. Нет ничего проще:

```python
squares={1:1, 2:4, 3:9}
```

```python
squares
```

```python
squares[1]
```

```python
squares[2]
```

В предыдущих двух строчках `squares` ведёт себя примерно как список,
но если внимательно приглядеться, то видно, что это не список, а всётаки словарь.


```python
squares
```

Например, у любого непустого списка есть элемент с индексом 0, а у
`squares` такого нет:

```python
squares[0] # происходит ошибка т.к. в словарях обращение к элементам идёт по ключу, а не по индексу
```

## Перебор записей в словаре
Как обрабатывать информацию в словаре? Для перебора всех элементов
списка можно было использовать цикл `for`. А что будет, если ему
скормить словарь вместо списка? Попробуем:


```python
for k in gradebook:
    print(k)
# перебираем каждый ключ из нашего словаря и выводим его на экран
```

Цикл `for` в этом случае перебирает все *ключи* нашего словаря. А зная
ключ, можно получить и значение:

```python
for k in gradebook:
    print(f'Студент {k} имеет оценку {gradebook[k]}')
# перебираем каждый ключь из нашего словаря и печатаем его и его значение 
```

Однако, есть более изящный способ получить сразу ключ и значение
очередной записи: использовать `items()`.

```python
for k, v in gradebook.items():
    print(f'Стуждент {k} имеет оценку {v}')
# теперь же мы перебираем и ключи и их значения, записывая их в соответствующие переменные и выводим их на экран
```

Как работает этот код? Здесь используется метод `items()`,
возвращающий список (точнее, итератор), состоящий из кортежей вида
`(ключ, значение)`.

```python
list(gradebook.items()) # получаем список, элементами которого являются кортежи из ключа и значения, соответствующего этому ключу
```

Оператор `for` в этом случае понимает, что нужно при каждом проходе
цикла выбрать очередной кортеж и присвоить его первый элемент (то
есть ключ) переменной `k`, а второй элемент (то есть значение)
переменной `v` (конечно, эти переменные могли бы называться иначе). С
аналогичным поведением мы уже встречались, когда обсуждали
конструкцию `enumerate`.

Вот так можно найти все записи с заданным значением — например,
всех студентов, получивших оценку 4:


```python
for k, v in gradebook.items():
    if v==4:
        print(k)
```

Заметим, что такой «поиск по значению» требует перебора всех записей
в словаре и если словарь большой, то он будет занимать много времени
— хотя «поиск по ключу» будет по-прежнему выполняться быстро.
Кстати, можно быстро проверить, существует ли в словаре запись с
данным ключом:


```python
"Коля" in gradebook # проверяем с помощью оператора in вхождение ключа в наш словарь
```

```python
"Alice" in gradebook
```

Если бы мы хотели искать среди значений, то нужно было бы явно это
указать с помощью метода `values()`:

```python
1 in gradebook.values() # проверяем вхождение значения в наш словарь с помощью метода .values()
```

```python
4 in gradebook.values()
```

Вообще оператор `in` не ограничивается только использованием со
словарями: он может использоваться, например, со списками:

```python
5 in [1, 2, 3, 5, 8] # проверяем вхождение 5 в список
```

```python
6 in range(1,5)      # проверяем вхождение 6 в диапозон
```

## Создание словарей и функция `zip()`
Есть разные способы создавать словари. Например, можно создать
пустой словарь и постепенно заполнять его элементами:

```python
my_dict = {}
```

```python
my_dict[1] = 1
my_dict['hello'] = 'world'
```

```python
my_dict
```

Заметим, что в одном и том же словаре прекрасно уживаются элементы
разных типов (в данном случае — строки и целые числа).
Можно создать словарь иначе, передав функции `dict()` список,
состоящий из пар ключ-значение (в некоторо смысл, это обратная
операция методу `items()`):


```python
dict([('hello', 'world'), ('one', 'two')]) # создаём словарь из двух кортежей, которые предсатавляют собой пары ключ-значение, с помощью функции dict()
```

Допустим, у нас есть два списка, в одном находятся имена студентов, а в
другом их оценки. Как можно из этих списков создать словарь, для
которого имена были бы ключами, а оценки значениями?

А вот так:

```python
students = ["Вася", "Коля", "Петя", "Аня"]
grades = [5, 4, 2, 3]
new_gradebook = list(zip(students,grades)) # "сшиваем" два списка в один с помощью функции zip(). 
                                           # Она устроена так: очередной элемент из первого списка связывается 
                                           # в кортеж с элементом того-же индекса второго, третьего (если есть) и т.д. списка
new_gradebook
```

Здесь используется удобная функция `zip()`, применение которой не
ограничивается созданием словарей. Подобно застёжки-молнии, она
«состёгивает» (отсюда и название) несколько списков. Например,
`zip()` делает из пары списков список пар:

```python
list(zip([1,2,3],['a','b','c']))
```

Эту конструкцию можно использовать, когда нам нужно перебрать
элементы двух связанных между собой списков. Например, вот так
можно вывести информацию о том, какой студент какую оценку имеет,
не используя словари:


```python
for student, grade in zip(students, grades): # создаём список из студентов и соответствующих им оценок и перебераем их, выводя на экран
    print(student, "has grade", grade)
```

Функцию `zip()` можно использовать и более чем с двумя списками:

```python
list(zip([1,2,3,4], [5,6,7,8], ['a','b','c','d']))
# списка три, поэтому на выходе получится список из троек
```

Если какой-то из списков окажется короче, то `zip()` «обрежет»
остальные списки:

```python
list(zip([1,2,3], ['a','b']))
```

```python
sums = {(2,3): 5, (4, 1): 5, (5, 7): 12}
```

```python
sums
```

```python
sums[(2, 3)]
```

```python
sums[(4, 1)]
```

```python
sums[(5, 7)]
```

В этом месте проявляется важное отличие кортежей от списков:
последние не могут быть ключами словарей, поскольку могут
изменяться.


```python
sums = { [1,2]: 3 }
```

На этом мы закончим краткое введение в словари и перейдём к
следующей теме.

# Списковые включения (list comprehensions)
Мы ранее частенько сталкивались с такой задачей: дан список, в котором
записаны числа, но в виде строчек. Создать новый список, в котором
числа были бы числами. Мы могли решить эту задачу с помощью цикла:

```python
str_list = ["1", "5", "12", "7"]
int_list = []
for s in str_list:
    int_list.append(int(s))

print(int_list)
```

За создание нового списка отвечают три строчки. Писать их каждый раз
довольно тоскливо, и создатели Python придумали (точнее, заимствовали
из функциональных языков программирования, а те его заимствовали у
математиков) гораздо более изящный синтаксис. Он устроен вот так:

```python
int_list = [int(s) for s in str_list] # перебираем каждый элемент из str_list и применяем функцию int() и записываем в новый список int_list
```

Кадратные скобки вокруг выражения должна подсказать, что мы создаём
список (потому что когда нужно создать список, мы обычно заключаем
его элементы в квадратные скобки). Выражение внутри скобок нужно
читать буквально:

*список, состоящий из элементов `int(s)` для ( `for` ) элементов `s` из ( `in` ) списка `str_list`*

```python
int_list
```

Кавычки исчезли — перед нами список, состоящий из чисел. Магия!
Исходный список `str_list` при этом не изменился:

```python
str_list
```

Аналогично можно применять любую операцию к элементам списка.
Например, возведём все элементы из `int_list` в квадрат:

```python
[x**2 for x in int_list] # перебираем каждый элемент из списка int_list, возводим в квадрат и записываем в список
```

или удвоим все элементы списка:

```python
double_list = [x*2 for x in int_list]
```

```python
double_list
```

или прибавим к ним 1:

```python
[x+1 for x in int_list]
```

или превратим их в числа с плавающей точкой:

```python
[float(x) for x in int_list]
```

Как видите, с элементами списков можно делать что угодно! Однако, это
ещё не все. В синтаксисе списочных включений можно производить
фильтрацию. Например, нам нужны только те элементы, которые больше
6. Мы можем их выбрать таким образом:


```python
[x for x in int_list if x>6] # перебираем каждый элемент из int_list и записываем его только если выполняется условие x>6
```

Когда мы пишем здесь `x for x` мы имеем в виду, что нужно просто
подставить в новый список элементы старого, ничего с ними не делая
(только выбирая нужные). Но можно и как-то их модифицировать:


```python
[x**2 for x in int_list if x>6]
```

Решим теперь такую задачу: есть два списка с числами, а мы хотим
найти их поэлементную сумму.


```python
X = [2, 5, 8]
Y = [1, 3, 100]
```

Её можно решить таким образом (для перебора элементов двух списков
одновременно используем конструкцию `zip()` , обсуждавшуюся выше):

```python
Z = []
for x, y in zip(X, Y):
    Z.append(x + y)
print(Z)
```

Но со списочными включениями тот же код выглядит гораздо
симпатичнее:


```python
[x + y for x, y in zip(X, Y)]
```

Кстати, можно использовать синтаксис, похожий на списочное
включение, чтобы создавать словари:


```python
squared = {i: i**2 for i in range(10)} # списковые включения также доступны и для словарей. Необходимо записать "ключ: значение" 
                                       # которые можно создавать с помощью списковых включений
squared
```

# Функция `map()`
У списочных включений есть аналог, который сейчас считается не
слишком удобным, но иногда встречается: функция `map()`.

```python
str_list
```

Вот так она решается с помощью `map()`:

```python
int_list = list(map(int, str_list)) # создаём список таким образом, что для каждого элемента из списка str_list применяем функцию int. Возможно это благодаря функции map()
```

```python
list(int_list)
```

Функция `map()` принимает два аргумента. Первым аргументом она
принимает функцию, после этого она применяет эту функцию к каждому
из элементов списка. В общем, записи вида `list(map(int, str_list))` и `[int(x) for x in str_list]` почти эквивалентны. Когда действие, которое нужно применить, уже существует в виде
функции (как в случае с `int` ) то конструкция с `map()` выглядит даже ... *элегантнее*

```python
[int(x)+1 for x in str_list]
```

Чтобы реализовать это с помощью `map()` , нужно объявить новую
функцию, которая будет возвращать значение выражения `int(x)+1` и
передать её `map()` .

```python
def my_func(x):
    return int(x)+1
```

```python
list(map(my_func, str_list))
```

Для краткости можно использовать `lambda`-функции, но такой подход
гораздо менее прозрачен, чем списочные включения, и сейчас
использовать его не рекомендуется.

## Два слова об эффективности
Использовать списочные включения не только приятно, но и полезно:
они работают эффективнее, чем код с циклом.


```python
from random import random
from math import sqrt

N = 10**4

mylist = [random() for _ in range(N)]
```

```python
%%timeit

newlist = []
for x in mylist:
    newlist.append(sqrt(x))
```

```python
%%timeit
newlist = [sqrt(x) for x in mylist]
```

```python
%%timeit
newlist = list(map(sqrt, mylist))
```

Как видно из этих данных (магическое слово `%timeit` позволяет
измерить, сколько времени уходит на какую-то операцию), списочные
включения быстрее обычного цикла. `map()` работает примерно с такой же скоростью, как и списочные включения (иногда чуть медленнее,
иногда чуть быстрее).

# Сложные структуры данных
Списки позволяет сохранить некоторый ряд значений, но зачастую
нужно уметь работать с более сложные структурами — например, с
таблицами. В некоторых языках программирования есть *двумерные массивы*. Аналогом двумерного массива в Python является «список
списков», то есть такой список, элементами которого являются другие
списки. С чем-то подобным мы уже встречались.

Рассмотрим пример: таблица, в которой записаны результаты по
нескольким домашним работам у нескольких студентов. (Допустим, мы
присвоили студентам некоторые номера и поэтому нам не нужно знать,
кого как зовут.) Её можно записать в виде списка списков, например, по
строчкам:


```python
table = [["HW1", "HW2", "HW3", "HW4"], [4, 3, 4, 4], [3, 4, 3, 2], [4, 4, 5, 4], [4, 5, 5, 3]] # создаём двумерный список каждый элемент которого - тоже список
```

Здесь каждый элемент списка `table` — это строчка нашей таблицы, то
есть тоже список. Например, вот так можно узнать, что записано в
третьей строке и четвертом столбце нашей таблицы:

```python
table[2][3]
```

Что здесь произошло? Мы сначала вызвали третью строку таблицы с
помощью


```python
table[2]
```

А потом из этой третьей строки выбрали четвертый элемент с помощью
`[3]` . Можно было бы записать это более подробно:


```python
row = table[2]
print(row[3])
# row[3] это то же самое, что table[2][3]
```

Вот так можно напечатать все элементы таблицы по строчкам:


```python
for row in table:
    print(*row)
```

```python
gradebook['Bob'][1]
```

1. Создайте словарь с числами от 1 до 10, где ключами будут числа, а
значениями — квадраты этих чисел.


```python
squares = {x: x**2 for x in range(1,11)}
print(squares)
```

2. Дана строка "verification". Создайте словарь, где ключами будут
буквы строки, а значениями – количество вхождений каждой буквы
«i» в строку.


```python
verify = {c: "verification".count("i") for c in "verification"}
print(verify)
```

3. Объедините два словаря в один (задания 1 и 2), используя
встроенные функции Python.

```python
new_dict = {**squares, **verify}
print(new_dict)
```

4.  Перемножьте все значения словаря (задания 1) и выведите результат
на экран.

```python
result = 1

for k in squares:
    result *= squares[k]

print(result)
```

8. Создайте список заглавных букв из слова «Парашют», используя
списочные включения (list comprehension).


```python
parashute = []
[parashute.append(c.upper()) for c in "Парашют" if c.upper() not in parashute]
print(parashute)
```

9. Сформируйте последовательность чисел на основе
предоставленного списка (список задать самостоятельно), удвоив
числа меньше 5 и отняв 2 от остальных.


```python
from random import randint

a = [randint(0,20) for _ in range(10)]
print(a)
print("-----------------------------------------")
result = [x*2 if x<5 else x-2 for x in a]
print(result)
```

10. Создайте функцию, которая принимает массив чисел и возвращает
новый массив, где каждый элемент умножен на 2. Используйте
функцию map для выполнения этой операции.


```python
a = [randint(0,20) for _ in range(10)]
print(a)
print("-----------------------------------------")

def double(n):
    return n*2

result = dict(map(double, a))
print(result)
```
