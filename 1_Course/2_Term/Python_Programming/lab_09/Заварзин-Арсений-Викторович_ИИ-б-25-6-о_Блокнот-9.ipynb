# Что такое объектно-ориентированное программирование?

До сих пор мы писали программы в *процедурном* стиле: данные
хранились в переменных, а функции работали с этими данными.
Объектно-ориентированное программирование (ООП) предлагает другой
подход: **объединить данные и функции, которые с ними работают, в один объект**.

Представьте, что вы хотите описать автомобиль. У него есть
характеристики (цвет, марка, скорость) — это **атрибуты** (данные). И есть
действия (завести двигатель, нажать на газ, затормозить) — это **методы**
(функции). В ООП мы создаём *класс* — чертёж, по которому создаются
конкретные *объекты* (экземпляры).

> Простой пример: в Python всё является объектом! Списки,
> строки, функции — все они имеют свои методы.
> `my_list.append(5)` — это вызов метода `append` у
> объекта-списка `my_list`


## Основные принципы ООП:
+ **Инкапсуляция** — сокрытие внутреннего устройства объекта.
+ **Наследование** — создание новых классов на основе существующих.
+ **Полиморфизм** — возможность работать с разными объектами через
единый интерфейс.

### Инкапсуляция
Инкапсуляция (от лат. in capsula — «внутри капсулы») — это принцип
объектно-ориентированного программирования, согласно которому
внутренние данные объекта (атрибуты) и методы их обработки
объединяются в единую сущность (класс), а доступ к этим данным извне
ограничивается или полностью скрывается. Внешний мир
взаимодействует с объектом только через открытый интерфейс — набор
публичных методов.

> «Инкапсуляция — это свойство языка, позволяющее
> объединить данные и методы, работающие с ними, в
> классе, а также скрыть детали реализации от пользователя,
> предоставив открытый интерфейс.»

> «В Python инкапсуляция не является строгой, как в
> некоторых других языках. Вместо этого она реализована на
> уровне соглашений: атрибуты и методы с одним
> подчёркиванием в начале (_attr) считаются
> «защищёнными», а с двумя подчёркиваниями (__attr) —
> «приватными». Однако это лишь соглашение, и технически
> доступ к таким атрибутам возможен.» — Официальная
> документация Python (PEP 8 — Style Guide for Python Code)

## Первый класс
Создадим класс, который представляет студента. У студента есть имя и
оценка. Он может представиться.

```python
class Student:
    """Класс, представляющий студента"""
    def __init__(self, name, grade):
        """Конструктор — вызывается при создании объекта"""
        self.name = name # атрибут объекта
        self.grade = grade # ещё один атрибут
        
    def introduce(self):
        """Метод — функция, принадлежащая объекту"""
        print(f"Меня зовут {self.name}, моя оценка — {self.grade}")
```

Разберём, что здесь происходит:
1. `class Student:` — объявление нового класса.
2. `def __init__(self, name, grade):` — специальный метод, **конструктор**. Он вызывается автоматически, когда мы создаём новый объект.
3. `self` — ссылка на сам объект. Через `self` мы обращаемся к атрибутам и методам объекта внутри класса.
4. `self.name = name` — создаём атрибут `name` у объекта и присваиваем ему значение.
5. `def introduce(self):` — обычный метод, который может использовать атрибуты объекта через `self` .

Теперь создадим конкретного студента — *экземпляр* класса `Student` :

```python
student1 = Student("Анна", 5)
student2 = Student("Иван", 4)
```

```python
student1.introduce()
student2.introduce()
```

Обратите внимание: при вызове `student1.introduce()` нам не нужно
передавать `self` — Python делает это автоматически. `student1`
становится `self` внутри метода.

К атрибутам объекта можно обращаться напрямую:

```python
print(student1.name)
print(student2.grade)
```

Можно менять атрибуты:

```python
student2.grade = 5
student2.introduce()
```

## Атрибуты класса и атрибуты экземпляра
Есть два вида атрибутов:
+ **Атрибуты экземпляра** — принадлежат конкретному объекту. Разные объекты могут иметь разные значения. (Мы уже видели `self.name` .)
+ **Атрибуты класса** — принадлежат самому классу и являются общими для всех объектов.


```python
class Student:
    university = "Севастопольский государственный университет" # Атрибут класса (общий для всех экземпляров)
    
    def __init__(self, name, grade):
        self.name = name # Атрибут экземпляра (уникальный для каждого студента)
        self.grade = grade # Атрибут экземпляра (уникальные для каждого студента)

    def introduce(self):
        print(f"{self.name} учится в {Student.university}, оценка: {self.grade}")
```

```python
s1 = Student("Мария", 4)
s2 = Student("Пётр", 5)
s1.introduce()
s2.introduce()
print(Student.university) # ваш комментарий
```

## Методы класса и статические методы
Помимо обычных методов (которые получают `self` ), есть ещё:
+ **Методы класса** — получают сам класс ( `cls` ) вместо экземпляра. Объявляются с помощью `@classmethod`.
+ **Статические методы** — не получают ни `self` , ни `cls`. Объявляются с помощью `@staticmethod` . Это просто функции, которые логически принадлежат классу.


```python
class MathOperations:
    pi = 3.14159
    
    @classmethod
    def circle_area(cls, radius):
        """Метод класса — может обращаться к атрибутам класса"""
        return cls.pi * radius ** 2
    
    @staticmethod
    def add(a, b):
        """Статический метод — не зависит от класса"""
        return a + b
```

```python
print(MathOperations.circle_area(5))
print(MathOperations.add(3, 7))
# можно вызывать и через экземпляр, но обычно так не делают
m = MathOperations()
print(m.add(10, 20))
```

## Специальные методы (магические методы)
В Python есть методы с двойным подчёркиванием в начале и конце. Они
позволяют определить поведение объектов для встроенных операций. Мы
уже знаем `__init__` . Рассмотрим другие полезные.



### `__str__` и `__repr__` — **строковое представление**

`__str__` вызывается, когда мы используем `print()` или `str()` .

`__repr__` — для более формального представления (вызывается в
интерпретаторе).


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"Точка ({self.x}, {self.y})"
    def __repr__(self):
        return f"Point({self.x}, {self.y})"

p = Point(3, 5)
print(p) # Вызов магического метода __str__() - возвращает "Точка (3, 5)" для удобного вывода пользователю
p # Вызов магического метода __repr__() - возвращает "Point(3, 5)" в интерактивной среде (джупитере)
```

### `__add__` , `__sub__` — арифметические операции

Можно определить, что значит «сложение» для наших объектов.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        """Перегрузка оператора +"""
        return Vector(self.x + other.x, self.y + other.y)
    
    def __sub__(self, other):
        """Перегрузка оператора -"""
        return Vector(self.x - other.x, self.y - other.y)
    
    def __str__(self):
        return f"({self.x}, {self.y})"
    
v1 = Vector(2, 3)
v2 = Vector(1, 4)
v3 = v1 + v2

print(v3) # Вызов магического метода __str__() - преобразует объект Vector в строку для вывода (3, 7)
print(v1 - v2) # Вызов магического метода __sub__() - выполняет вычитание векторов и возвращает новый Vector (1, -1)
```

### `__len__` и `__getitem__` — поведение, похожее на список
Можно сделать так, чтобы объект вёл себя как последовательность.

```python
class Group:
    def __init__(self, students):
        self.students = students

    def __len__(self):
        return len(self.students)
    
    def __getitem__(self, index):
        return self.students[index]

group = Group(["Анна", "Иван", "Мария", "Пётр"])
print(len(group)) # Вызов магического метода __len__() - возвращает количество студентов (4)
print(group[1]) # Вызов магического метода __getitem__() - получение элемента по индексу (Иван)
for s in group: # Итерация по объекту благодаря __getitem__() - делает group итерируемым объектом
    print(s)
```

## Наследование
Наследование — это механизм объектно-ориентированного
программирования, позволяющий создать новый класс (производный,
дочерний, подкласс) на основе уже существующего класса (базового,
родительского, суперкласса). Дочерний класс автоматически получает все атрибуты и методы родительского класса и может дополнять их новыми
или переопределять существующие. Наследование выражает отношение
«является» (is-a) между объектами.

> «Python поддерживает множественное наследование. Класс
> может наследоваться от нескольких базовых классов. При
> поиске атрибута интерпретатор Python обходит дерево
> наследования в порядке, определяемом алгоритмом C3
> (Method Resolution Order).» — Официальная документация
> Python

**Наследование** позволяет создать новый класс на основе существующего.
Новый класс (дочерний) получает все атрибуты и методы родительского и
может их расширять или изменять.

Представим, что у нас есть общий класс `Animal` , а от него мы создадим
`Dog` и `Cat` .


```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def make_sound(self):
        print("Какое-то животное издаёт звук")
    
    def move(self):
        print(f"{self.name} двигается")

class Dog(Animal): # Dog наследуется от Animal
    def make_sound(self): # Переопределение метода родительского класса - полиморфизм
        print(f"{self.name} говорит: Гав-гав!")
    
    def fetch(self): # Новый метод, специфичный только для класса Dog
        print(f"{self.name} приносит палку")

class Cat(Animal):
    def make_sound(self): # Переопределение метода родительского класса - полиморфизм
        print(f"{self.name} говорит: Мяу-мяу!")
    
    def climb(self): # Новый метод, специфичный только для класса Cat
        print(f"{self.name} залезает на дерево")
```

```python
rex = Dog("Рекс")
murka = Cat("Мурка")
rex.make_sound() # Полиморфный вызов - будет выполнен метод make_sound() из класса Cat (Гав-гав!)
murka.make_sound() # Полиморфный вызов - будет выполнен метод make_sound() из класса Cat (Мяу-мяу!)
rex.move() # Вызов унаследованного метода из родительского класса Animal (если он там определен)
rex.fetch() # Вызов уникального метода класса Dog, отсутствующего в Animal
murka.climb() # Вызов уникального метода класса Dog, отсутствующего в Animal
```

### Функция `super()`
Если в дочернем классе мы переопределяем `__init__`, но хотим
выполнить код родительского конструктора, используется `super()`.

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def info(self):
        print(f"Студент: {self.name}, оценка: {self.grade}")

class GraduateStudent(Student):
    def __init__(self, name, grade, thesis_title):
        super().__init__(name, grade) # используем super(), ведь мы переопределяем __init__
        self.thesis_title = thesis_title

    def info(self):
        super().info() # вызывает метод info() из родительского класса, расширяя его функциональность в дочернем классе
        print(f"Тема курсовой работы: {self.thesis_title}")

gs = GraduateStudent("Алексей", 5, "Машинное обучение")
gs.info()
```

## Инкапсуляция
**Инкапсуляция** — это сокрытие внутреннего состояния объекта от
внешнего вмешательства. В Python нет строгой приватности, но есть
соглашения:

+ `_atribute` — «защищённый» (protected) — намекает, что не стоит обращаться напрямую.
+ `__atribute` — «приватный» (private) — Python изменяет имя (name mangling), чтобы случайно не обратиться.

Лучше использовать **геттеры и сеттеры** (методы для доступа к
атрибутам) или свойство `@property`.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance # создаём поле "баланс" приватным

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Внесено {amount}. Баланс: {self.__balance}")
        else:
            print("Сумма должна быть положительной")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Снято {amount}. Баланс: {self.__balance}")
        else:
            print("Недостаточно средств или неверная сумма")

    def get_balance(self): # делаем функцию для получения баланса, ведь иначе мы его не получим
        return self.__balance

account = BankAccount("Иван", 1000)
print(account.get_balance()) # вместо .__balance вызываем метод .get_balance(), который выдаёт нам баланс
account.deposit(500)
account.withdraw(300)
```

```python
print(account.__balance) # мы не можем получить баланс, ведь он является "приватным"
```

### `@property` — элегантные геттеры и сеттеры
Декоратор `@property` позволяет обращаться к методу как к атрибуту.

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            print("Температура не может быть ниже абсолютного нуля!")
        else:
            self._celsius = value

    @property
    def fahrenheit(self):
        return self._celsius * 9/5 + 32

t = Temperature(25)
print(t.celsius) # обращаемся к методу как к атрибуту
print(t.fahrenheit) # обращаемся к методу как к атрибуту
t.celsius = 30 # обращаемся к сеттеру celsius и задаём знчаение 30
print(t.celsius) # снова обращаемся к методу как к атрибуту
t.celsius = -300 # обращаемся к сеттеру и ловим ошибку, ведь -300 не попадает под условие >= -273.15
```

## Полиморфизм
**Полиморфизм** — возможность работать с разными объектами через
единый интерфейс.

Полиморфизм (от греч. polys — «много» и morphe — «форма») — это
принцип объектно-ориентированного программирования, позволяющий
объектам разных классов выполнять вызовы методов по-своему, в
соответствии с их собственной реализацией. Полиморфизм означает, что
один и тот же интерфейс может использоваться для работы с объектами
различных типов, а конкретное поведение определяется типом объекта во
время выполнения программы.

> «В Python полиморфизм проявляется особенно ярко
> благодаря динамической типизации. Любая функция может
> работать с любым объектом, если этот объект поддерживает
> ожидаемые операции. Это неформальное правило часто
> называют «утиной типизацией» (duck typing): «Если что-то
> выглядит как утка, плавает как утка и крякает как утка, то
> это, вероятно, утка, и с ним можно работать как с уткой».»
> — Официальная документация Python

```python
class Duck:
    def sound(self):
        return "Кря-кря!"
    
    def move(self):
        return "Плаваю"
    
class Dog:
    def sound(self):
        return "Гав-гав!"
    
    def move(self):
        return "Бегаю"
    
class Car:
    def sound(self):
        return "Бип-бип!"
    
    def move(self):
        return "Еду"
    
def describe(obj):
    """
    Функция работает с любым объектом, у которого есть
    sound() и move()
    """
    print(f"Звук: {obj.sound()}")
    print(f"Перемещение: {obj.move()}")

describe(Duck())
print("---")
describe(Dog())
print("---")
describe(Car())
```

## Абстрактные классы
Иногда мы хотим создать класс, который нельзя инстанцировать (создать
объект), но который задаёт интерфейс для дочерних классов. Для этого
используются абстрактные классы из модуля `abc`.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    """Абстрактный класс — нельзя создать объект"""
    
    @abstractmethod
    def area(self):
        """Должен быть реализован в дочерних классах"""
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)
    
rect = Rectangle(5, 3)
print(f"Площадь: {rect.area()}, Периметр: {rect.perimeter()}")
```

```python
shape = Shape() # Shape - является абстрактным классом, потому его нельзя инициализировать таким образом
```

## Композиция против наследования
Не всегда нужно использовать наследование. Иногда лучше использовать
**композицию** — когда один объект содержит другой объект в качестве
атрибута.


```python
class Engine:
    def start(self):
        print("Двигатель запущен")
    
    def stop(self):
        print("Двигатель остановлен")

class Wheels:
    def rotate(self):
        print("Колёса вращаются")

class Car:
    """Используем композицию: Car имеет Engine и Wheels"""
    
    def __init__(self):
        self.engine = Engine()
        self.wheels = Wheels()
    
    def drive(self):
        self.engine.start()
        self.wheels.rotate()
        print("Машина едет!")
    
    def stop(self):
        self.engine.stop()
        print("Машина остановлена")
        
car = Car()
car.drive()
car.stop()
```

**Правило:** Используйте наследование для отношения «является» ( `is-a` ),
композицию для отношения «имеет» ( `has-a` ). Собака *является*
животным → наследование. Машина *имеет* двигатель → композиция.

2. Создайте класс `Library` , который хранит список книг. Реализуйте
методы `add_book()` (добавить книгу), `remove_book()` (удалить
по названию) и `find_by_author()` (найти все книги автора).

```python
class Library:
    def __init__(self):
        self.books = []
    
    def addBook(self, title, author, year):
        book = Book(title, author, year)
        self.books.append(book)
        print(f'Книга "{title}" добавлена в библиотеку!')
    
    def removeBook(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                print(f'Книга "{title}" удалена!')
                return True
        print(f'Книга "{title}" не найдена!')
        return False
    
    def findByAuthor(self, author):
        found_books = []
        for book in self.books:
            if book.author == author:
                found_books.append(book)
        
        if found_books:
            print(f"\nКниги автора '{author}':")
            for book in found_books:
                book.info()
                print()
        else:
            print(f"Книги автора '{author}' не найдены!")
        
        return found_books
    



library = Library()
library.addBook("Война и мир", "Лев Толстой", 1869)
library.addBook("Преступление и наказание", "Федор Достоевский", 1866)
library.addBook("Анна Каренина", "Лев Толстой", 1877)
library.addBook("Евгений Онегин", "Александр Пушкин", 1833)

# Ищем книги по автору
library.findByAuthor("Лев Толстой")

# Удаляем книгу
library.removeBook("Преступление и наказание")
```

4. Создайте класс `Fraction` для работы с обыкновенными дробями.
Реализуйте методы `__add__` , `__sub__` , `__mul__` , `__truediv__` ,
`__str__` . Дроби должны сокращаться.

```python
import math

class Fraction:
    def __init__(self, ch, zn):
        if zn == 0:
            print("Знаменатель не должен быть равен нулю!")
            return

        self.ch = ch
        self.zn = zn
        self.reduce()
    
    def reduce(self):
        gcd = math.gcd(abs(self.ch), abs(self.zn))
        self.ch //= gcd
        self.zn //= gcd

    def __add__(self, other):
        if isinstance(other, Fraction):
            new_ch = self.ch * other.zn + other.ch * self.zn
            new_zn = self.zn * other.zn
            return Fraction(new_ch, new_zn)
        elif isinstance(other, (int, float)):
            # Преобразуем целое число в дробь
            return self + Fraction(other, 1)
        else:
            print("Ошибка создания дроби!")
            return self
        
    def __sub__(self, other):
        if isinstance(other, Fraction):
            # a/b - c/d = (a*d - c*b) / (b*d)
            new_ch = self.ch * other.zn - other.ch * self.zn
            new_zn = self.zn * other.zn
            return Fraction(new_ch, new_zn)
        elif isinstance(other, (int, float)):
            return self - Fraction(other, 1)
        else:
            print("Ошибка сложения дробей!")
            return self
        
    def __mul__(self, other):
        if isinstance(other, Fraction):
            # (a/b) * (c/d) = (a*c) / (b*d)
            new_ch = self.ch * other.ch
            new_zn = self.zn * other.zn
            return Fraction(new_ch, new_zn)
        elif isinstance(other, (int, float)):
            return self * Fraction(other, 1)
        else:
            print("Ошибка умножения дробей!")
            return self
        
    def __truediv__(self, other):
        if isinstance(other, Fraction):
            # (a/b) / (c/d) = (a*d) / (b*c)
            new_ch = self.ch * other.zn
            new_zn = self.zn * other.ch
            return Fraction(new_ch, new_zn)
        elif isinstance(other, (int, float)):
            return self / Fraction(other, 1)
        else:
            print("Ошибка деления дробей!")
            return self
        
    def __str__(self):
        if self.ch == 0:
            return "0"
        elif self.zn == 1:
            return str(self.ch)
        else:
            return f"{self.ch}/{self.zn}"

f1 = Fraction(1, 2)
f2 = Fraction(1, 3)
f3 = Fraction(6, 9)

print("Создание дробей:")
print(f"f1 = {f1}")
print(f"f2 = {f2}")
print(f"f3 = {f3} (сокращена)")
print()

print("Сложение:")
print(f"{f1} + {f2} = {f1 + f2}")
print(f"{f1} + 2 = {f1 + 2}")
print()

print("Вычитание:")
print(f"{f1} - {f2} = {f1 - f2}")
print(f"{f1} - 2 = {f1 - 2}")
print()

print("Умножение:")
print(f"{f1} * {f2} = {f1 * f2}")
print(f"{f1} * 3 = {f1 * 3}")
print()

print("Деление:")
print(f"{f1} / {f2} = {f1 / f2}")
print(f"{f1} / 2 = {f1 / 2}")
print()
```

6. Создайте класс `BankAccount` с приватным атрибутом `__balance` .
Реализуйте методы `deposit()` , `withdraw()` и свойство `balance`
(только для чтения).


```python
class BankAccount:
    def __init__(self, balance):
        self.id = id(self)
        self.__balance = balance

    def deposit(self, count):
        self.__balance += count
        print(f"Счёт пополнен на {count} руб! Баланс: {self.__balance}")

    @property
    def balance(self):
        return self.__balance

    def withdraw(self, count):
        if self.__balance < count:
            print("На счёте недостаточно средств!")
            return
        self.__balance -= count
        print(f"Снято {count} руб! Баланс: {self.__balance}")

myAcc = BankAccount(100000)
myAcc.withdraw(500000)
myAcc.deposit(10)
print(f'Ваш баланс: {myAcc.balance}')
myAcc.withdraw(10000)
print(f'Ваш баланс: {myAcc.balance}')
```

7. Создайте абстрактный класс `Vehicle` с абстрактными методами
`start()` и `stop()` . Создайте классы `Car` и `Bicycle` ,
реализующие эти методы.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def __init__(self):
        pass

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def __init__(self, color):
        self.color = color
    
    def start(self):
        print(f"{self.color} машина поехала! Тапку в пол!!!")

    def stop(self):
        print(f"{self.color} машина остановилась! Ставлю на ручник.")

class Bicycle(Vehicle):
    def __init__(self, color):
        self.color = color

    def start(self):
        print(f"{self.color} велосипед поехал! Кручу педали!")

    def stop(self):
        print(f"{self.color} велосипед остановился! Пристёгиваю к забору.")

myCar = Car('Красная')
myBike = Bicycle('Серый')

myCar.start()
myBike.start()
myBike.stop()
myCar.stop()
```

8. Создайте класс `Student` с атрибутами `name` , `grades` (список
оценок). Добавьте свойство `average` (средний балл) и метод
`add_grade()` .

```python
class Student:
    def __init__(self, name, grades=None):
        self.name = name
        self._grades = grades if grades is not None else []
    
    @property
    def grades(self):
        return self._grades.copy()  # Возвращаем копию для защиты данных
    
    @property
    def average(self):
        if not self._grades:
            return 0.0
        return sum(self._grades) / len(self._grades)
    
    def add_grade(self, grade):
        self._grades.append(grade)
    
    def __str__(self):
        return f"Студент: {self.name}, Средний балл: {self.average:.2f}"

# Создание студента
student1 = Student("Иван Петров")
print(student1)
print(f"Средний балл: {student1.average}")

# Добавление оценок
student1.add_grade(90)
student1.add_grade(100)
student1.add_grade(89)
student1.add_grade(100)

print(f"\nПосле добавления оценок: {student1}")
print(f"Список оценок: {student1.grades}")
print(f"Средний балл: {student1.average:.2f}")

# Создание студента с начальными оценками
student2 = Student("Мария Сидорова", [5, 5, 4, 5])
print(f"\n{student2}")
print(f"Средний балл: {student2.average:.2f}")
```

9. Создайте класс `Time` для работы со временем (часы, минуты,
секунды). Реализуйте метод `__add__` для сложения двух моментов
времени и метод `__str__` для вывода в формате "ЧЧ:ММ:СС".

```python
class Time:
    def __init__(self, hour, min, sec):
        if 0 <= hour <= 23:
            self.hour = hour
        else:
            self.hour = 0
            
        if 0 <= min <= 59:
            self.min = min
        else:
            self.min = 0
            
        if 0 <= sec <= 59:
            self.sec = sec
        else:
            self.sec = 0

    def __str__(self):
        return f'{self.hour:02}:{self.min:02}:{self.sec:02}'
    
    def to_sec(self):
        return self.hour * 3600 + self.min * 60 + self.sec
    
    @classmethod
    def from_sec(cls, total_sec):
        total_sec = total_sec % (24 * 3600)
        hour = total_sec // 3600
        min = (total_sec % 3600) // 60
        sec = total_sec % 60
        return cls(hour, min, sec)

    def __add__(self, other):
        total_sec = self.to_sec() + other.to_sec()
        return Time.from_sec(total_sec)

t1 = Time(1, 12, 50)        
t2 = Time(4, 12, 10)

print(t1)  # 01:12:50
tNew = t1 + t2
print(tNew)  # 05:25:00
```

10.  Создайте класс `Bank` с атрибутом класса `interest_rate`
(процентная ставка). Реализуйте метод класса
`set_interest_rate()` для изменения ставки для всех счетов.

```python
class Bank:
    interest_rate = 0.05
    
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.balance = balance
    
    @classmethod
    def set_interest_rate(cls, new_rate):
        if new_rate < 0:
            raise ValueError("Процентная ставка не может быть отрицательной")
        cls.interest_rate = new_rate
        print(f"Процентная ставка изменена на {new_rate * 100}%")

    @classmethod
    def get_interest_rate(cls):
        return cls.interest_rate

    def __str__(self):
        return f"Владелец: {self.account_holder}, Баланс: {self.balance:.2f} руб."
    
r1 = Bank('Resident 1', 10000)
r2 = Bank('Resident 2', 2000)
r3 = Bank('Resident 3', 10000000)

print(r1)

Bank.set_interest_rate(0.25)

print("Ставка сейчас ->", Bank.get_interest_rate())

print(r2)
```
