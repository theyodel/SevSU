 # Списки и операции с ними
 Список - это такая структура данных, которая содержит в себе сразу много элементов.

````python
numbers = [4, 8, 9, 2, 6] # создаём список с именем numbers и элементами 4, 8, 9, 2, 6
````

````python
numbers # выводим на экран список numbers
````

 В списках можно хранить не только числа. Например, создадим список из строк.

````python
strings = ["Hello", "World", "Test"] # создаём список с именем strings и значениями "Hello", "World", "Test"
````

 В списке могут храниться данные разных типов. Например, строки, целые числа, числа с плавающей точкой.

````python
mixed_list = ["Hello", 6, 7.8] # создаём список с названием mixed_list и значениями разных типов 
````

 Можно обращаться к отдельным элементам списка и работать с ними как с обычными переменными. Чтобы выбрать элемент нужно указать его номер.

````python
print(numbers) # выводим список numbers на экран
print(numbers[1]) # выводим элемент с индексом 1 на экран
````

**Нумерация начинается с нуля!** На самом деле, у этого правила есть свои [рациональные обоснования](http://pythonhistory.blogspot.ru/2013/10/why-python-uses-0-based-indexing.html)



 > Гвидо ван Россум, создатель Python, поясняет, что выбор индексации с нуля был продиктован не столько наследием языка C, сколько элегантностью и удобством полуоткрытых интервалов в сочетании со срезами.



Если в списке есть элементы разных типов, они никак не «мешают» друг другу. Например, наличие в списке строк не превращает другие элементы этого списка в строки.



 > Это касается только обычных списков Python. Несколько позже мы будем проходить массивы numpy и там всё не так.

````python
print(mixed_list) # выводим список mixed_list на экран
print(mixed_list[2]+4) # прибавляем к элементу с индексом 2 значение 4 и выводим на экран
````

 Элементы списка можно менять так же, как значения обычных переменных.

````python
numbers # выводим на экран список numbers
````

````python
numbers[1] = 222 # присваиваем элементу с индексом 1 значение 222
numbers          # выводим на экран список numbers
````

> Списки Python основаны на стандартных C'шных массивах и обладают их свойствами с точки зрения производительности: в частности, обращение к элементу по его индексу имеет сложность $O(1)$, то есть не является массовой операцией

Чтобы узнать длину списка, можно использовать функцю `len`.

````python
len(numbers) # выводим на экран длину (кол-во элементов) списка numbers
````

Заметим, что это не индекс последнего элемента, а именно число элементов. Если вам нужно получить последний элемент то его индексом будет `len(numbers)-1`. Но в Python можно обращаться к элементам списка, считая их «с конца», гораздо проще:

````python
numbers = [4, 8, 2, 5] #задаём новые значения для списка numbers
numbers[-1] # выводим элемент с индексом -1 (первый с конца)
````

````python
numbers[-2]
````

 А вот если вы попытаетесь обратиться к элементу с несуществующим индексом, то получите ошибку.

````python
numbers[5] = 100 # присваиваем элементу с индексом 5 значение 100, но ловим ошибку т.к. не существует элемента с индексом 5
````

 Однако дописывать элементы в конец можно:

````python
numbers = [7, 6, 2]     # создаём список numbers
print(numbers)          # выводим numbers на экран
numbers.append(777)     # добавляем 777 в конец списка numbers
print(numbers)          # выводим получившийся список numbers
````

Слово `append` — это так называемый «метод» — функция,
«принадлежащая» некоторому объекту (в данном случае — объекту
`numbers` типа `list` (список)), и что-то делающая с этим объектом.
У `numbers`, как у любого списка, есть много методов. Можно
набрать `numbers`. , нажать табуляцию (после точки), и получить
список доступных методов. А ещё можно набрать `help(list)` или
даже `help(numbers)` (в нашем случае) и получить краткое описание
этих методов. Например, так можно узнать, что помимо append у
списков есть метод `extend`.


````python
print(numbers)              # печатаем numbers
numbers.extend([3, 7, 5])   # добавляем в конец numbers 3 элемента
print(numbers)              # выводим изменённый список numbers
````

Метод `extend` позволяет приписать к списку сразу несколько
элементов. Он получает на вход список, который нужно приписать:
обратите внимание на квадратные скобки внутри круглых при вызове
этого метода — они создают новый список, который и передаётся
функции `extend`. Методы `append` и `extend` меняют список, к
которому они применяются. Иногда вместо этого нужно создать новый
список, объединив (конкатенировав!) два других. Это тоже можно
сделать.


````python
first_list = [5, 8, 2]      # создаём первый список
second_list = [1, 9, 4]     # создаём второй список
new_list = first_list + second_list #создаём третий лист из двух предыдущих
print(new_list)     # выводим результат на экран
````

Плюсик в данном случае обозначает не поэлементное сложение (как вы могли подумать), а конкатенацию. Cписки `first_list` и `second_list` при этом не изменились

````python
print(first_list)
print(second_list)
````

У вас могло возникнуть желание использовать сложение вместо операции `extend`.

````python
print(numbers)
# не надо так
numbers = numbers + [2, 6, 9]
print(numbers)
````

Вообще говоря, этот код сработал, но делать так не следует: при выполнении операции конкатенации создаётся новый список, затем в него копируются все элементы из `numbers`, потом к ним приписываются элементы из второго списка, после чего старый `numbers` забывается. Если бы в `numbers` было много элементов, их копирование в новый список заняло бы много времени. Гораздо быстрее приписать элементы к уже готовому списку.

## Срезы

Иногда нам нужен не весь список, а его кусочек. Его можно получить, указав в квадратных скобках не одно число, а два, разделённых двоеточием.

````python
print(numbers)
print(numbers[1:4]) # срез из элементов с индексом 1 2 и 3
````

Это называется *slice* (по-русски часто говорят  *срез*). Обратите внимание: левый конец среза включается (элемент с индексом 1 — это шестёрка), а правый — нет. Так будет всегда. Это соглашение оказывается удобным, например, потому что позволяет посчитать число элементов в срезе — нужно из правого конца вычесть левый (в данном случае 4-1=3).

Если левый элемент не указан, то он считается началом списка, а если правый — то концом.

````python
print(numbers[7:]) # срез из элементов, начиная с 7 и до конца
print(numbers[:7]) # срез из элементов начиная с 0 и до 6 (по индексам)
````

Всегда верно следующее: список `numbers` это то же самое, что `numbers[:k]+numbers[k:]`, где `k` — любой индекс.

Срезы можно использовать для присваивания.

````python
numbers = [5, 8, 9, 10]
print(numbers[1:3])
numbers[1:3]= [55, 77]
print(numbers)
````

Не обязательно, чтобы список, который мы присваиваем срезу, имел ту же длину, что и срез. Можно присвоить более длинный список (тогда исходный список расширится), а можно менее длинный (тогда сузится). Можно использовать срезы, чтобы вставить несколько элементов внутрь списка. (Для одного элемента это можно делать с помощью метода `insert`.)

````python
numbers = [6, 8, 9]
print(numbers[1:1])         # это пустой срез
numbers[1:1] = [99, 77, 55] # добавляем вместо элемента с индексом 1 сразу 3 элемента (99 77 55)
print(numbers)
````

> Чтобы вставить какие-то элементы внутрь списка, необходимо освободить для него место, сдвинув все последующие элементы вперёд. Python сделает это автоматически, но время это займёт. Поэтому, по возможности, следует этого избегать, особенно если вы работаете с большими массивами данных. Если вам очень нужно записывать что-нибудь в начало и конец списка, посмотрите на двустороннюю очередь (deque) из модуля `collections`.

Можно удалять элементы списка (и вообще что угодно) или срезы с помощью команды `del`.

````python
numbers = [6, 7, 9, 12, 8, 3]
del(numbers[4]) # удалим 8
print(numbers)
del(numbers[0:2]) # удалим 6 7
print(numbers)
````

# Практические задачи

1)  Пользователь вводит числа до тех пор, пока не введет слово «finish». Поместите введенные числа в список. Если сумма элементов списка больше 50, то добавляем в конец списка «5».

````python
a = []                      # вводим список, куда будем записывать числа
flag = input()              # вводим флаг, по которому будем определять новые числа или слово finish
while flag != "finish":     # бесконечный цикл для ввода цифр пока не достигнем финиша
    a.append(float(flag))   # переводим флаг в float (если это не финиш) и добавляем его в список
    flag = input()          # считываем новый флаг

if sum(a)>50:   # проверяем сумму и если она >50, то дописываем в список число 5.
    a.append(5)

print(a) # печатаем список
````

2) Пользователь вводит числа до тех пор, пока не введет слово «finish». Поместите введенные числа в список. Если сумма элементов списка меньше 50, то к списку добавляем список [2, 3, 4, 5].

````python
a = []
flag = input()
while flag != "finish":
    a.append(float(flag))
    flag = input()

if sum(a)<50:           # проверяем сумму и если она <50, то дописываем в список [2,3,4,5]
    a.extend([2,3,4,5])

print(a)
````

3) Пользователь вводит числа до тех пор, пока не введет слово «finish». Поместите введенные числа в список. Если сумма элементов по модулю списка больше 100, то к списку добавляем список [4, 3, 5, 6]. Иначе добавляем в конец списка «3».

````python
import math     # импортируем модуль математики (можно с помощью from math import *)
a = []
flag = input()
while flag != "finish":
    a.append(float(flag))
    flag = input()

if math.fabs(sum(a))>100:   # проверяем модуль суммы и если он >100, то дописываем в список [4,3,5,6]
    a.extend([4,3,5,6])
else:                       # иначе 3
    a.append(3)

print(a)
````

4) Пользователь вводит числа до тех пор, пока не введет слово «finish». Поместите введенные числа в список. Если размер списка больше последнего элемента списка, то необходимо отсортировать список по возрастанию.

````python
a = []
flag = input()
while flag != "finish":
    a.append(float(flag))
    flag = input()

if len(a)!=0 and a[-1]<len(a):  # сравниваем последний элемент с длиной списка. len(a)!=0 проверяет наличие хотябы 1 элемента в списке, чтобы не выбивало ошибку
    a.sort()                    # сортируем список по возрастанию (если в .sort() указать reverse=False, то будет по убыванию)

print(a)
````

5) Пользователь вводит числа до тех пор, пока не введет слово «finish». Поместите введенные числа в список. Если размер списка меньше предпоследнего элемента списка, то необходимо отсортировать список по убыванию.

````python
a = []
flag = input()
while flag != "finish":
    a.append(float(flag))
    flag = input()

if len(a)>=2 and a[-2]>len(a): # сравниваем предпоследний элемент с длиной списка. len(a)>=2 проверяет наличие хотябы 2 элементов в списке, чтобы не выбивало ошибку
    a.sort(reverse=False)

print(a)
    
````

````python
a = [0]*10          # создаём список из 10 элементов, каждый из которых равен 0
for i in range(10): # перебираем и записываем в каждый элемент списка некоторе число с клавиатуры. Вместо 10 в range() можно было написать len(a)
    a[i] = int(input(f'Введите {i+1} элемент: '))

for i in range(9):  # указано 9 т.к. если написать 10, то программа будет искать 11 элемент, что выдаст ошибку.
    if a[i]<a[i+1]: # проверяем соседние элементы на условие
        print(a[i+1], end="\t") # печатаем подходящие под условие элементы в строку (можно убрать end="\t" чтобы элементы печатались в столбик)
````

7) Пользователь вводит 15 чисел с клавиатуры. Необходимо внести их в список. Поменять местами первый и последний элементы списка.

````python
a = [0]*15
for i in range(15):
    a[i] = int(input(f'Введите {i+1} элемент: '))

temp = a[0]     # записываем первый элемент в temp
a[0] = a[-1]    # записываем вместо первого элемента последний
a[-1] = temp    # в последний элемент записываем значение первого из переменной temp

print(a)
````

8) Пользователь вводит 15 чисел с клавиатуры. Необходимо внести их в список. Поменять местами максимальный и минимальный элементы списка.

````python
a = [0]*15
for i in range(15):
    a[i] = int(input(f'Введите {i+1} элемент: '))

temp = a[a.index(max(a))]                 # записываем в temp значение максимального элемента
a[a.index(max(a))] = a[a.index(min(a))]   # записываем вместо максимального элемента минимальный
a[a.index(min(a))] = temp                 # на место минимального элемента ставим значение максимального из temp

print(a)
````

12) Дано два списка из 10 элементов. Необходимо рассчитать суммы для каждого из списков. Если сумма первого списка больше чем сумма первого, то необходимо объединить списки.

````python
a, b = [0]*10, [0]*10 # вводим два списка по 10 элементов
for i in range(10):
    a[i] = randint(-100, 100)
    b[i] = randint(-100, 100)

print(a, b)

if sum(a) > sum(b): # проверяем суммы элементов двух списков
    a.extend(b)     # если условие верно, то дописываем к списку a список b

print("\nОтвет:", a) # выводим ответ
````

13) Дано два списка из 10 элементов. Необходимо рассчитать суммы отрицательных элементов для каждого из списков. Если сумма элементов первого списка меньше суммы второго, то необходимо добавить в конец каждого списка «1».

````python
a, b = [0]*10, [0]*10
for i in range(10):
    a[i] = randint(-100, 100)
    b[i] = randint(-100, 100)

print(a, b)
sumA, sumB = 0, 0       # вводим переменные для хранение сумм отрицательных элементов списокв a и b

for i in range(10):     # перебираем оба массива и складываем отрицательные элементы
    if a[i]<0:
        sumA += a[i]
    if b[i]<0:
        sumB += b[i]

if sumA < sumB:
    a.append(1), b.append(1)

print("\nОтвет:", a, b)
````

14) Дано три списка, числа которых вводятся с клавиатуры до слова «end». Необходимо найти произведение для каждого из списков. Если сумма произведений списков больше 100, то необходимо отсортировать каждый список по возрастанию.

````python
a, b, c = [], [], [] # вводим 3 пустых списка
flag = input()
while flag != "end":
    a.append(int(flag))
    flag = input()
print("Список a записан!")

flag = input()
while flag != "end":
    b.append(int(flag))
    flag = input()
print("Список b записан!")

flag = input()
while flag != "end":
    c.append(int(flag))
    flag = input()
print("Список c записан!")

if len(a)==0 or len(b)==0 or len(c)==0: # проверяем каждый массив на наличие хотябы 1 элемента
    print("Один из списков пустой! Выполнение программы невозможно!")
else:
    pA, pB, pC = 1, 1, 1 # переменные для сохранения произведений элементов каждого массива
    for i in range(len(a)):     # перемножаем элементы каждого списка
        pA *= a[i]
    for i in range(len(b)):
        pB *= b[i]
    for i in range(len(c)):
        pC *= c[i]
    if pA+pB+pC > 100:
        a.sort()
        b.sort()
        c.sort()
        print("\nОтвет:", a, b, c)
    else:
        print("\nОтвет:", a, b, c)
````

15) Дано три списка, числа которых вводятся с клавиатуры до слова «end». Необходимо найти произведение для каждого из списков. Если сумма произведений списков меньше 290, то необходимо отсортировать каждый список по убыванию.

````python
a, b, c = [], [], [] # вводим 3 пустых списка
flag = input()
while flag != "end":
    a.append(int(flag))
    flag = input()
print("Список a записан!")

flag = input()
while flag != "end":
    b.append(int(flag))
    flag = input()
print("Список b записан!")

flag = input()
while flag != "end":
    c.append(int(flag))
    flag = input()
print("Список c записан!")

if len(a)==0 or len(b)==0 or len(c)==0: # проверяем каждый массив на наличие хотябы 1 элемента
    print("Один из списков пустой! Выполнение программы невозможно!")
else:
    pA, pB, pC = 1, 1, 1 # переменные для сохранения произведений элементов каждого массива
    for i in range(len(a)):     # перемножаем элементы каждого списка
        pA *= a[i]
    for i in range(len(b)):
        pB *= b[i]
    for i in range(len(c)):
        pC *= c[i]
    if pA+pB+pC < 290:
        a.sort(reverse=False)
        b.sort(reverse=False)
        c.sort(reverse=False)
        print("\nОтвет:", a, b, c)
    else:
        print("\nОтвет:", a, b, c)
````

    b[i] = i

2. Как работает индексация в списках Python? Что означает
отрицательный индекс (например, `numbers[-1]`) и почему попытка обратиться к элементу numbers[5] в списке из 4 элементов вызовет ошибку?
> **Ответ:** Индексация в списке из `n` элементов в Python начинается с `0` и заканчивается `n-1`. Отрицательный индекс начинается с `-1` и идёт до `-n`. Это обозначает что мы идём от конца списка. При попытке обратиться к элементу `numbers[5]` в списке из `4` элементов обернётся ошибкой т.к. индекс `5` выходит за границы допустимого и вызывает ошибку `IndexError`.

> **Ответ:** Синтаксис `numbers[начало:конец]` возвращает новую последовательность, содержащую элементы, начиная с индекса `начало` **включительно** и до индекса `конец` **не включительно**.
